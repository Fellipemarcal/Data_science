```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('amazon_sales_dataset.csv') 
```

```python
cat_rev = (
    df
    .groupby('product_category')['total_revenue']
    .sum()
    .sort_values(ascending = False)
)
```

```python
top_cat = cat_rev.sort_values(ascending = False).head(3).values
```

```python
cat_rev[cat_rev.isin(top_cat)]
```
