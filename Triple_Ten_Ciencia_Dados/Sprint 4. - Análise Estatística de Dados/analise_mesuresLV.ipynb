```python
import pandas as pd
import numpy as np
from collections import Counter
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('social_media_addiction.csv')
```

```python
df.columns
```

```python
'''Pourquoi nous utilisons pas mediane dans ce cas?
Parce que ces données possèdent pas une distribution assimétrique,
donc utiliser moyenne est la meilleur option dans ce moment


donc la moyenne d'utilisation é 132.5'''


np.mean(df['Daily_Usage_Time_min'])
```

```python
print(Counter(df['Platform']))
```

```python
'''Il faut savoir que quand .mode() trouve des elements avec la meme quantité
il nous retourne les données triés avec l'ordre alphabetique

Interpretation:
Facebook, Instagram et Tiktok sont les plus utilisé parmi nos données !

Autres Points Importants:
Pouvons-nous appliquer le teoreme de Simpson pour voir la realité ? ->
Pour découvrir si quelque chose se cache derrière analyse Global des données ? '''

df['Platform'].mode()
```

```python
max(df['Daily_Usage_Time_min']) - min(df['Daily_Usage_Time_min'])
```

```python
'''Rappelons-nous de la moyenne >>> 135.5'''

'''np.var, nous retourne 5648.75, maintenant interessons-nous à l'ecart_type
dans cas passons à utiliser np.std

np.std, nous retourne 75.15, voici notre écart-type

avec ces mesures en mains, nous découvrons que nos données ont une grande variance

135,5 + 75,15 = 210,65
135,5 - 75,15 = 60,35

Ça signifie que nos données peuvent varier entre 60,35 à 210,65, concluons que les données
de la colonne 'Daily_Usage_Time_min' sont très dispersés
'''
np.var(df['Daily_Usage_Time_min']) #5648.75
np.std(df['Daily_Usage_Time_min']) #75.15
```
