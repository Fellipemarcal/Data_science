```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('social_media_addiction.csv')
```

```python
df.columns
```

```python
df['Age'].agg(['min','max','mean'])
```

```python
tmp_plt = (
    df
    .groupby('Platform')['Daily_Usage_Time_min']
    .sum()
)
plt.figure(figsize = (10,5))
plt.title("Temps d'usage sur chaque plateform")
plt.ylabel('Temps en minutes')
ax = tmp_plt.plot(kind = 'bar' )
for i in ax.containers:
    ax.bar_label(i)
plt.xticks(rotation = 0)
plt.show()
```

```python
plt.figure(figsize = (10,5))
tmp_plt.plot(kind = 'pie',autopct = '%i%%')
plt.ylabel("")
plt.title('Utilisation des Plateformes')
plt.show()
```
