```python
import pandas as pd
```

```python
age = [10,19,20,13,11,24,25,66]
nom = ['a','b','c','d','e','f','g','h']

df = pd.DataFrame({
    'nom':nom,
    'age':age
})
```

```python
df
```

```python
def categorie_age(age):
    if age < 18:
        return "mineur"
    elif age < 60:
        return "adulte"
    else:
        return "senior"

df['status'] = df['age'].apply(categorie_age)
```
