```python
import pandas as pd
```

```python
#Traitement des données avant l'execution de l'exemple
df = pd.read_csv('data.csv')
df['tax'] = df['tax'].fillna(df['tax'].mode()[0])
df['gender'] = df['gender'].fillna('inconnu')
df['country'] = df['country'].fillna('inconnu')
df['price_paid'] = (
    df['price_paid']
    .astype(str)
    .str.replace('$','', regex = False)
    .str.replace(',','', regex = False)
    .astype(float)
)
```

```python
df
```

```python
def prod_cher(i): # les fonctions sont lues de l'haut en bas
    if i < 4: # dans ce cas nous devonts mettres les valeurs plus petit toujours dans l'haut du code
        return 'pas cher'
    elif i < 7 :
        return 'cher'
    else:
        return 'très cher'
    
```

```python
df['classe_prod'] = df['price_paid'].apply(prod_cher) # utilisatation de la function 'prod_cher' sur 'price_paid' avec .apply
#on crée une nouvelle colonne pour strocker le résultat
```

```python
df
```

```python
df["groupe_prodt"] = pd.cut( #Ici nous emploions pd.cut à la place de .apply (il faut pd.cut nous contraint à créer une nouvelle colonnne))
    df["price_paid"], #nom de la colonne
    bins=[0, 4, 7, 10], #difinitions des valeurs (0, min, centre, max)
    labels=["A", "B", "C"] #les valeurs de 0 à 4  = A; 4 à 7 = B; 7 à 10 = C                          
)
```
