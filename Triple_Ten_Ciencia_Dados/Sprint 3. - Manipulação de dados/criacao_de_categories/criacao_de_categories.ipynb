```python
import pandas as pd
```

```python
nom = ['Fellipe', 'Joao','Maria', 'Fernanda']
age = [21,40,33,60]
rev = [19000,13000,6000, 9000]
df = pd.DataFrame({
    'nom': nom,
    'age': age,
    'rev': rev
})
```

```python
def profil_client(row): # row=ligne
    if row["rev"] > 5000 and row["age"] < 30:
        return "jeune_rich"
    else:
        return "standard"

df["profil"] = df.apply(profil_client, axis=1)
    
```
