```python
df = pd.read_csv('bug_dataset_50k.csv')
```

```python
df.isnull().sum()
```

```python
mask = (
    (df['severity'] == 'Medium') &
    (df['bug_category'] == 'API Bug') &
    (df['bug_domain'] == 'Cloud') &
    (df['tech_stack'] == 'PostgreSQL')
)
```

```python
mode_value = df.loc[mask, 'error_code'].mode()[0]
```

```python
df.loc[mask & df['error_code'].isna(), 'error_code'] = mode_value
```
