```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('social_media_addiction.csv')
```

```python
def cls_age(i): #IL faut changer ce code 
    if i < 26.5:
        return 'Plus Jeune'
    else:
        return 'Adulte'
df['Classe_age'] = df['Age'].apply(cls_age)
```

```python
df.columns
```

```python
plt_gn = (
    df
    .groupby(['Platform','Gender'])['Daily_Usage_Time_min']
    .sum()
    .sort_values(ascending = False)
)
plt_gn
```

```python
plt_gn_ut = (
    df
    .groupby('Gender')['Daily_Usage_Time_min']
    .sum()
    .sort_values(ascending = False)
)
plt_gn_ut
```

```python
#utilisation par age (division par classe)
age_ut = (
    df
    .groupby('Classe_age')['Daily_Usage_Time_min']
    .sum()
    .sort_values(ascending = False)
)
age_ut
#nous puvons aussi reaiser autres analyses, tels comme : 'Quel plateforme est la plus populaire
#entre dans chaque groupe-> c'est le meme principe qu'on a utilisé avec pour découvrir
#quelle plateform est la plus utilisé par chaque genre
```

```python
#quelle plateforme est la plus populaire dans chaque groupe ?
age_plt = (
    df
    .groupby(['Platform','Classe_age'])['Daily_Usage_Time_min']
    .sum()
    .sort_values(ascending = False)
)
age_plt

#jeunes -> TikTok
#adultes -> Facebook
```

```python
age_plt = (
    df
    .groupby('Classe_age')['Platform']
    .sum()
    .sort_values(ascending = False)
)
age_plt
```
