```python
from matplotlib import pyplot as plt
variance = [1, 2, 4, 8, 16, 32, 64, 128, 256]
bias_square = [256, 128, 64, 32, 16, 8, 4, 2, 1]
total_error = [x + y for x, y in zip(variance, bias_square)]
xs = [i for i, _ in enumerate(variance)]
```

```python
#nous pouvons multipliquer les appels à plt.plot
#pour montrer plusieurs series sur le même graphique
plt.plot(xs, variance, 'g-', label='variance')
plt.plot(xs, bias_square, 'r-', label='bias^2')
plt.plot(xs, total_error, 'b:', label='total error')
#parce que nous avons assigné des labels pour chaque serie7
#nous avons une légende gratuite
#loc=9 veut dire "trop center" (centré en haut)
plt.legend(loc=9)
plt.xlabel("Complexité du modèle")
plt.title("Le compromis entre biais et variance")
plt.show()
```
